# 03 — Exploratory Data Analysis (EDA) Walkthrough

Let's visualize catalog growth, top countries & genres, audience maturity across decades, and our extra deep-dive charts.

```python
%matplotlib inline
from netflix_analysis.data_loader import load_netflix_data
from netflix_analysis.cleaning import clean_netflix_data
from netflix_analysis.preprocessing import preprocess_netflix_data
from netflix_analysis.eda import (
    plot_growth_trajectory,
    plot_top_countries,
    plot_country_genre_heatmap,
    plot_ratings_by_decade,
)

raw = load_netflix_data()
prep = preprocess_netflix_data(clean_netflix_data(raw.df).df)
```

```python
c1 = plot_growth_trajectory(prep.df)
c1.figure
```

```python
c2 = plot_top_countries(prep.titles_by_country)
c2.figure
```

```python
c3 = plot_country_genre_heatmap(prep.titles_by_country, prep.titles_by_genre)
c3.figure
```

```python
c4 = plot_ratings_by_decade(prep.df)
c4.figure
```
