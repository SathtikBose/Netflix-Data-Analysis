Here we turn the cleaned dataset into analysis-ready tables:
1. Separate `duration` into `Movie_Duration_Min` and `TV_Seasons_Count`.
2. Compute `Addition_Lag_Years` and `Catalog_Category`.
3. Explode multi-label `listed_in` (genres) and `country` columns.

```python
from netflix_analysis.data_loader import load_netflix_data
from netflix_analysis.cleaning import clean_netflix_data
from netflix_analysis.preprocessing import preprocess_netflix_data

raw = load_netflix_data()
cleaned = clean_netflix_data(raw.df)
prep = preprocess_netflix_data(cleaned.df)
prep.df[['title', 'type', 'Movie_Duration_Min', 'TV_Seasons_Count', 'Addition_Lag_Years', 'Catalog_Category']].head(5)
```
